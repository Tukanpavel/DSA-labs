2. Рассчитайте выборочное среднее и стандартное отклонение для выборки.
3. Постройте 95% доверительный интервал для среднего роста в генеральной
совокупности, считая стандартное отклонение известным (8 см).
4. Постройте 95% доверительный интервал, считая стандартное отклонение
неизвестным (используя t-распределение).
5. Сравните ширину интервалов.
